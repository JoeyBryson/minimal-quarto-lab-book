```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def plot_sine_wave(T):
    x = np.linspace(-2, 2, 1000)
    y = np.tanh(4*(1/T)*x)

    plt.figure(figsize=(8, 4))
    plt.plot(x, y, label=r"$\tanh(4 \beta \langle \sigma_i \rangle )$")
    plt.plot(x, x, label=r"$\langle \sigma_i \rangle$")
    plt.legend()
    plt.xlabel(r"$\langle \sigma_i \rangle$")
    plt.ylabel(r"$\tanh(4 \beta \langle \sigma_i \rangle )$")
    plt.grid(True)
    plt.title("figure 1")
    plt.show()

interact(plot_sine_wave, T=widgets.FloatSlider(min=0.1, max=10, step=0.1, value=1.0))
```
