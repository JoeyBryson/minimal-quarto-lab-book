Graph titles should be 'Binder Cumulant'

Given large samples of data of magnetization, the Binder Cumulant can be calculated from the distributions the standard formula

$$
U_N = 1 - \frac{\langle \sigma^4 \rangle}{3\langle \sigma^2 \rangle^ 2}
$$

A variety of methods were then used estimate intersections 

First, polynomial fits were attempted. And the intersection points were then solved for each pair of polynomials. This tended to produce innacuraturate fits implying degree being too low but then also instability and overfitting when the degree was increased. Additionally the intersection points, which were averaged to be averaged to estimate the critical temperature had many false intersections caused by the additional occilations induced by the polynomial fits. Which heavily shifted the estimate away from the true critical temperature as can be seen from the graphs

Another method that was used was spline fitting. Which benifitted from more accurately representing local behaviour without being skewed by global behaviour. Which is important when determining a precice intersection point.

the above graph has rendering artifacts, the spline functions found were continuous.

A method that was tried to avoid the issues with intersection points was to define a parameter which quantifies how close the spline functions are to eachother.
This was effectively the reciprocal of the variance of the spline functions at a given temperature, and then finding the peak of this parameter. Although variance may be a missleading term to use in this context, it gave useful estimates of the critical temperature. It was noted in hindsight that this parameter was probably dominated by the furthest apart spline functions.
