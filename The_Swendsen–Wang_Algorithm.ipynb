## The Swendsen–Wang Algorithm

Up to this point in the lab, the algorithm used to evolve the system into an equilibrium state, and then cycle through enough states to collect a representative sample of the distributions of Energy and Magnetization, was the Metropolis-Hastings Algorithm. This algorithm chose a single cell at random and flipped it with a probability according it's own spin state and the spin state of it's nearest neighbours. This is possible because the Hamiltonian defining the Ising Model:

$H = -J \sum_{\langle i j \rangle} S_i S_j.$

depends only on the interactions between the cells i and j which are defined by the notation $\langle i j \rangle$ to be adjacent. Therefore, the metropolis algothithm can accurately reproduce this hamiltonian by only considering the local energy change of flipping a single cell.

### The Problem of Critical Slowing Down

The Metropolis-Hastings algorithm can theoretically produce representative statistics for the Ising model, given a sufficiently long amount of simulation time. However, as the temperature approaches the critical temperature, the amount of time required for the system to reach an equlibrium state increases assymtotically. This effect is known as Critical Slowing Down, and given the practical limits of computation, can severely reduce the accuracy (relative to the theoretical limit) of data generated. Simulations were run over for a simulation time of 1000 equalization steps and 10000 measuremement steps.
