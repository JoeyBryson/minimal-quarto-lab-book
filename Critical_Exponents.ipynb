# Finding The Critical Exponents
## Finding $\nu$

First, simulations were run using the Swendsen-Wang implementation over a range of temperatures $2\le T \le 3$ and system sizes $8\le N \le 256$ for 1000 equilibrium steps and 10,000 measurement steps. The total magnetization and the total energy(the Hamiltonian) were saved at every time step and compiled into distributions. The suceptibility could then be calculated from the distribution using the standard formula given in the theory section. This entire process was repeated three times to give an estimate for error on each datapoint.

above the raw succeptibility data is presented showing assymtotic behaviour

It was found that using a polynomial fit to find the peak produced large errors on the peak position. A Savitzky–Golay filter was instead selected, to make the peak finding alogythm more accurate. This method had some important benifits such as preserving peak hight effectively. 

The error on the maximum position was estimated by 'bootstrap' method which involved running the analysis process multiple times with gausian noise added to with raw data with standard deviation equal to the error in the data (divided by square root three since we want the error on the mean).

The critical exponent can be derived directly using the relationship

$$
\lvert T_c(N^2) - T_c\rvert \propto N^{-1/\nu}
$$

so 

$$
T_c(N^2) = T_c + A N^{-1/\nu}
$$

By taking the peak position vs system size data, scipy.optimize.curve_fit was used to find the parameters $T_c, A, \nu$ of the above equation. 

This yielded a values of $\nu = 1.00\pm0.07\quad T_c = 2.2742\pm 0.0006$ and $A = 1.7\pm0.5$. This method produced large errors since T_c was presumed unknown, so a log-log plot was not posible. 

Above, the determined parameters were then plotted on a log-log plot to verify the relationship. It was noted that letting curve_fit find a value of T_c itself provided a much closer fit and a more linear relationship in the log-log regime than using our T_c found through the Binder Cumulant method. This is hypothesised to be because of divergence from ideal behaviour at low system sizes, skewing the fit.
