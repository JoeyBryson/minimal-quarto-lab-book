A small investigation was made into finding the correlation length $\xi(T)$ directly from it's definition

$$
\langle (\sigma_i - \langle \sigma_i \rangle)(\sigma_j -  \langle \sigma_j \rangle) \rangle \sim e^{-\frac{|x - y|}{\xi(T)}}
$$

first an N = 32 simulation was run for 4000 time steps. This would normally be a very quick operation, but in this investigation analysis was done as the simulation ran. The correlations between every pair of points in the lattice was calculated using many, averaged Kroniker Products of an array matrix representing the spins subtract the system magnetization mean(as in the definition above), with itself. A corresponding matrix of distance between every pair of points in the same format as the Kroneker Product Matrix was generated.
