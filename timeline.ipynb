30/ 01/ 2025

Simulations were ran over a range of temperatures for simulation sizes N = 4 to N = 128. From these simulations, the natural distributions of Magnetization and Energy could be extracted. Then, the Binder Cumulant, which parameterizes the distributions based on how previlent the tails of the distribution are (with a value of 0 corresponding to a gaussian), could be assigned to each simulation. By plotting the Binder Cumulant against Temperature for each simulation size, a curve can be fitted showing a slow decrease in the cumulant as temperature is increased. The principle is that for larger simulation sizes the curve drops off sharper because the distribution is closer to the thermodynamic limit which would be a perfectly sharp jump. The temperature at which this jump occurs is the Critical temperature. By comparing thhe curves for different simulation sizes, it could be seen that they all cross this critical temperature at the same point and this is a demonstration of scale invarience, since every simulation is equivilently insufficient to reach the limit of homoginiety, it doesn't exist!
